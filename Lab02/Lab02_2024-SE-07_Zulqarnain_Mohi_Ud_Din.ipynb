## Task 1 · Setup
Import NumPy as `np` and print its version.

```python
# Task 1: setup (numpy is already imported in the data cell; repeated here so the task stands alone)
import numpy as np
print("T1: NumPy version =", np.__version__)

def tidy(keys, vals, nd=2):
    """Pair labels with rounded values as plain Python types, so dict printouts stay readable."""
    return dict(zip(np.asarray(keys).tolist(), np.round(vals, nd).tolist()))
```

## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# Task 3: randomness and reproducibility
np.random.seed(2026)
u = np.random.rand()                            # (a) one uniform number in [0, 1)
print("T3a one uniform value:", round(u, 6))

z25 = np.random.randn(25)                       # (b) continues the same seeded stream
print("T3b 25 standard-normal values:\n", np.round(z25, 3))
print("T3b sample mean = %.4f, sample std = %.4f" % (z25.mean(), z25.std()))

np.random.seed(42)                              # (c) fresh seed for the dice
rolls = np.random.randint(1, 7, size=10_000)    # upper bound is excluded
faces, counts = np.unique(rolls, return_counts=True)
print("T3c faces :", faces)
print("T3c counts:", counts, "| total =", counts.sum())

p_six = (rolls == 6).mean()                     # (d) mean of a boolean array = fraction of True
print("T3d P(six) estimated = %.4f, theory 1/6 = %.4f, gap = %.4f" % (p_six, 1/6, abs(p_six - 1/6)))

is_six = rolls == 6                             # (e) a six followed immediately by another six
double_six = np.sum(is_six[:-1] & is_six[1:])
print("T3e consecutive double-sixes:", double_six, "| expected about", round(9_999 / 36, 1))
```

Using `units`, `pop_2023` and `pop_2017`:
- **(a)** Total population.

- **(c)** The most populous unit.
- **(d)** Units with more than 20 million people.
- **(e)** Annual growth rate 2017–2023 for every unit, in one line: `((pop_2023 / pop_2017) ** (1/6) - 1) * 100`.
- **(f)** The fastest-growing unit.

**Interpret:** in 1–2 lines, what do (e) and (f) say? Compare AJK with the rest.

```python
# Task 6: Census 2023
total = pop_2023.sum()                                   # (a)
print("T6a total population 2023: {:,} people".format(total))

share = pop_2023 / total * 100                           # (b) broadcasting: 7 values / 1 scalar
print("T6b share of total (%):", tidy(units, share))
print("T6b shares add up to: %.2f %%" % share.sum())

print("T6c most populous unit:", units[pop_2023.argmax()])           # (c)

big = pop_2023 > 20_000_000                              # (d) boolean mask
print("T6d units above 20 million people:", units[big])

growth = ((pop_2023 / pop_2017) ** (1 / 6) - 1) * 100   # (e)
print("T6e annual growth (% per year):", tidy(units, growth))

print("T6f fastest-growing: %s (%.2f %% per year)" % (units[growth.argmax()], growth.max()))   # (f)
print("T6 extra: AJK growth %.2f %% vs mean of the other six %.2f %%" % (
    growth[units == "AJK"][0], growth[units != "AJK"].mean()))
print("T6 extra: slowest-growing unit:", units[growth.argmin()])
```

## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# Task 7: climate of five cities
print("T7a shape of tmax:", tmax.shape, "(5 cities x 12 months)")                    # (a)

annual_mean = tmax.mean(axis=1)                                                      # (b) one value per city
print("T7b annual mean max temperature (degC):", tidy(cities, annual_mean))

hot_idx = tmax.argmax(axis=1)                                                        # (c)
print("T7c hottest month per city:", dict(zip(cities.tolist(), months[hot_idx].tolist())))
print("T7c ...with its temperature (degC):", tidy(cities, tmax.max(axis=1), 1))

monthly_avg = tmax.mean(axis=0)                                                      # (d) one value per month
print("T7d 5-city monthly average (degC):", tidy(months, monthly_avg))

hot_months = (tmax > 35).sum(axis=1)                                                 # (e)
print("T7e months above 35 degC:", tidy(cities, hot_months, 0))

lahore_f = tmax[0] * 9 / 5 + 32                                                      # (f)
print("T7f Lahore in degF:", np.round(lahore_f, 1))

monsoon = rain[:, 6:9].sum(axis=1)                                                   # (g) Jul, Aug, Sep
print("T7g monsoon rain Jul-Sep (mm):", tidy(cities, monsoon, 1))

print("T7h wettest monsoon city:", cities[monsoon.argmax()], "(%.1f mm)" % monsoon.max())   # (h)
monsoon_pct = monsoon / rain.sum(axis=1) * 100
print("T7h yearly rain (mm):", tidy(cities, rain.sum(axis=1), 1))
print("T7h monsoon share of yearly rain (%):", tidy(cities, monsoon_pct, 1))
```

**Interpretation (T7):** Karachi has the highest yearly mean maximum (32.51 °C) yet the lowest peak of all five cities (35.7 °C in May), which surprised me; the sea keeps it warm all year without extreme summer heat, whereas Peshawar reaches 39.6 °C in June. Quetta was the other surprise: only 6.0 % of its rain falls in the monsoon, because its rain arrives in winter.

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# Task 8: inflation FY2024 vs FY2025
avg = cpi.mean(axis=1)                                                               # (a)
print("T8a average inflation FY2024: %.2f %% | FY2025: %.2f %%" % (avg[0], avg[1]))

high24 = cpi[0] > 25                                                                 # (b)
print("T8b FY2024 months above 25 %:", cpi_months[high24], "->", cpi[0][high24], "%")

drop = cpi[0] - cpi[1]                                                               # (c)
print("T8c drop per month (percentage points):", tidy(cpi_months, drop, 1))

k = drop.argmax()                                                                    # (d)
print("T8d biggest drop: %s with %.1f percentage points (%.1f %% -> %.1f %%)" % (cpi_months[k], drop[k], cpi[0][k], cpi[1][k]))

label = np.where(cpi[1] < 5, "Low", np.where(cpi[1] < 10, "Moderate", "High"))      # (e)
print("T8e FY2025 labels:", dict(zip(cpi_months.tolist(), label.tolist())))
```

**Interpretation (T8):** Average year-on-year inflation fell from 26.22 % in FY2024 to 4.87 % in FY2025, a drop of 17 to 26 percentage points in every month (largest in January, 25.9 points). Prices were still rising, but far more slowly: from November onwards inflation was in the "Low" band and reached 0.3 % in April.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# Task 9: grace marks
after = np.clip(marks + 5, 0, 100)                                                   # (a)
print("T9a original marks:", marks)
print("T9a after grace   :", after)

pass_pct = (after >= 50).mean() * 100                                                # (b)
print("T9b pass rate after grace: %.1f %% (before grace: %.1f %%)" % (pass_pct, (marks >= 50).mean() * 100))

only_grace = np.flatnonzero((marks < 50) & (after >= 50))                            # (c)
print("T9c passed only because of grace -> indices:", only_grace, "| marks before:", marks[only_grace])

z = (after - after.mean()) / after.std()                                             # (d)
print("T9d z-scores:", np.round(z, 2))
print("T9d check -> mean = %.2f, std = %.2f" % (z.mean(), z.std()))
```

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
# Task 10: an image is an array
img = np.zeros((8, 8), dtype=int)                                                    # (a)
img[1, 1:7] = 255          # top bar of the 7
img[2:7, 5] = 255          # slanted stem, drawn as a vertical stroke
img[4, 3:7] = 128          # grey middle bar (overwrites pixel [4,5])
print("T10a image:\n", img)

flipped = img[:, ::-1]                                                               # (b)
print("T10b flipped left-right:\n", flipped)

inverted = 255 - img                                                                 # (c)
print("T10c inverted:\n", inverted)

crop = img[1:5, 3:7]                                                                 # (d) rows 1-4, cols 3-6
print("T10d crop (shape %s):\n" % (crop.shape,), crop)

bright = np.clip(img + 80, 0, 255)                                                   # (e)
print("T10e brightened by 80:\n", bright)

print("T10f shape:", img.shape, "| dtype:", img.dtype, "| white pixels (=255):", (img == 255).sum())   # (f)
```

```python
# Optional: see the image and the climate table
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 4, figsize=(13, 3.2))
for a, pic, t in zip(ax, [img, flipped, inverted, bright], ["original", "flipped", "inverted", "brightened +80"]):
    a.imshow(pic, cmap="gray", vmin=0, vmax=255); a.set_title(t); a.axis("off")
plt.tight_layout(); plt.show()
```
